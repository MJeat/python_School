# First exercise



```python
# Adjacent List

letter_list = {
    'A': ['B'],
    'B': ['C'],
    'C': ['E'],
    'D': ['B'],
    'E': ['D', 'F'],
    'F': []
}

print("===Adjacent List Representation===")
for first, second in letter_list.items():
    print(f"{first}: {second}")
```

```python
nodes = ['A', 'B', 'C', 'D', 'E', 'F']

# Adjacency matrix: 0=no edge, 1=edge exists (directed)
graph_adjacency_matrix = [
    [0, 1, 0, 0, 0, 0],  # A
    [0, 0, 1, 0, 0, 0],  # B
    [0, 0, 0, 0, 1, 0],  # C
    [0, 1, 0, 0, 0, 0],  # D
    [0, 0, 0, 1, 0, 1],  # E
    [0, 0, 0, 0, 0, 0]   # F
]

print("\n=== Adjacency Matrix Representation ===")
print("            ", end="")
for node in nodes:
    print(f"{node:8}", end="")
print()
for i, node in enumerate(nodes):
    print(f"{node:8}", end="")
    for j in range(len(nodes)):
        print(f"{graph_adjacency_matrix[i][j]:8}", end="")
    print()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Create a directed graph from adjacency list
G = nx.DiGraph(letter_list)

# Basic graph information
print("\nNumber of nodes:", G.number_of_nodes())
print("Number of edges:", G.number_of_edges())
print("\nNodes:", list(G.nodes()))
print("Edges:", list(G.edges()))

# Visualize the graph
plt.figure(figsize=(5, 4))
nx.draw(G, with_labels=True, node_color='lightblue', 
        node_size=1500, font_size=12, font_weight='bold',
        edge_color='gray', width=2, arrows=True, arrowsize=20)
plt.title("Directed Graph Visualization")
plt.show()
```

```python
# Adjacent List
second_list = {
    'Seattle': {'Boston': 160, 'Dallas': 230, 'Texas': 130},
    'Boston': {'Texas': 120},
    'Texas': {'Dallas': 125, 'Chicago': 120},
    'Dallas': {'Chicago': 300},
    'Chicago': {'Seattle': 500}
}


print("===Adjacent List Representation===")
for first, second in second_list.items():
    print(f"{first}: {second}")

```

```python
# Adjacent Matrix
nodes = ['Seattle', 'Boston', 'Texas', 'Dallas', 'Chicago']

# Adjacency matrix: 0=no edge, distance=edge weight (directed)
adj_matrix = [
    [0, 160, 130, 230, 0],    # Seattle
    [0, 0, 120, 0, 0],        # Boston
    [0, 0, 0, 125, 120],      # Texas
    [0, 0, 0, 0, 300],        # Dallas
    [500, 0, 0, 0, 0]         # Chicago
]

print("\n=== Adjacency Matrix Representation ===")
print("            ", end="")
for node in nodes:
    print(f"{node:8}", end="")
print()

for i, node in enumerate(nodes):
    print(f"{node:8}", end="")
    for j in range(len(nodes)):
        print(f"{adj_matrix[i][j]:8}", end="")  # Use adj_matrix here
    print()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Directed graph adjacency list with weights (distances)
second_list = {
    'Seattle': {'Boston': 160, 'Texas': 130, 'Dallas': 230},
    'Boston': {'Texas': 120},
    'Texas': {'Dallas': 125, 'Chicago': 120},
    'Dallas': {'Chicago': 300},
    'Chicago': {'Seattle': 500}
}

# Create undirected graph with weights
G = nx.Graph()

# Add edges along with weights (distance)
for node, edges in second_list.items():
    for adjacent_node, distance in edges.items():
        G.add_edge(node, adjacent_node, weight=distance) # adding distances in the nested loop using loop from the second_list

# Basic info
print("\nNumber of nodes:", G.number_of_nodes())
print("Number of edges:", G.number_of_edges())
print("\nNodes:", list(G.nodes()))
print("Edges:", list(G.edges(data=True)))  # shows weights

# Draw graph
plt.figure(figsize=(7, 5))
pos = nx.spring_layout(G)  # positions for all nodes

# Draw nodes and edges
nx.draw(G, pos, with_labels=True, node_color='lightblue',
        node_size=1500, font_size=12, font_weight='bold',
        edge_color='gray', width=2, arrows=True, arrowsize=20)

# Draw edge labels (weights)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.title("Directed Graph with Distances")
plt.show()
```
