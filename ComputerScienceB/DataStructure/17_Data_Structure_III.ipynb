#### 1. Adjacency List (using dictionary)

```python
# Adjacency List Representation
# Example: Social Network

graph_adjacency_list = {
    'Alice': ['Bob', 'Charlie', 'David'],
    'Bob': ['Alice', 'Charlie'],
    'Charlie': ['Alice', 'Bob', 'David', 'Emma'],
    'David': ['Alice', 'Charlie'],
    'Emma': ['Charlie']
}

print("Adjacency List Representation:")
for person, friends in graph_adjacency_list.items():
    print(f"{person}: {friends}")
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Create an undirected graph
G = nx.Graph(graph_adjacency_list)

# Basic graph information
print("Number of nodes:", G.number_of_nodes())
print("Number of edges:", G.number_of_edges())
print("\nNodes:", G.nodes())
print("Edges:", G.edges())

# Visualize the graph
plt.figure(figsize=(4, 3))
nx.draw(G, with_labels=True, node_color='lightblue', 
        node_size=1500, font_size=12, font_weight='bold',
        edge_color='gray', width=2)
plt.title("Social Network Graph")
plt.show()

# You can also perform various graph operations
print("\nDegree of each person:")
print(dict(G.degree()))

print("\nShortest path from Alice to Emma:")
print(nx.shortest_path(G, 'Alice', 'Emma'))
```

#### 2. Adjacency Matrix (using 2D list)

```python
# Adjacency Matrix Representation
# 0 = no edge, 1 = edge exists

# Nodes: Alice(0), Bob(1), Charlie(2), David(3), Emma(4)
graph_adjacency_matrix = [
    [0, 1, 1, 1, 0],  # Alice
    [1, 0, 1, 0, 0],  # Bob
    [1, 1, 0, 1, 1],  # Charlie
    [1, 0, 1, 0, 0],  # David
    [0, 0, 1, 0, 0]   # Emma
]

nodes = ['Alice', 'Bob', 'Charlie', 'David', 'Emma']

print("Adjacency Matrix Representation:")
print("     ", end="")
for node in nodes:
    print(f"{node:8}", end="")
print()

for i, node in enumerate(nodes):
    print(f"{node:8}", end="")
    for j in range(len(nodes)):
        print(f"{graph_adjacency_matrix[i][j]:8}", end="")
    print()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Create an undirected graph
G = nx.Graph()
G.add_nodes_from(nodes)
for i in range(len(nodes)):
    for j in range(i+1, len(nodes)):  # Only upper triangle to avoid duplicates
        if graph_adjacency_matrix[i][j] == 1:
            G.add_edge(nodes[i], nodes[j])

# Basic graph information
print("Number of nodes:", G.number_of_nodes())
print("Number of edges:", G.number_of_edges())
print("\nNodes:", G.nodes())
print("Edges:", G.edges())

# Visualize the graph
plt.figure(figsize=(4, 3))
nx.draw(G, with_labels=True, node_color='lightblue', 
        node_size=1500, font_size=12, font_weight='bold',
        edge_color='gray', width=2)
plt.title("Social Network Graph")
plt.show()

# You can also perform various graph operations
print("\nDegree of each person:")
print(dict(G.degree()))

print("\nShortest path from Alice to Emma:")
print(nx.shortest_path(G, 'Alice', 'Emma'))
```

#### 2. Depth-First Search (DFS)

**Concept**: Explore as far as possible along each branch before backtracking

**Implementation**: Using recursion or stack

```python
# DFS Implementation (Recursive)

def dfs_recursive(graph, start, visited=None):
    """
    Perform Depth-First Search on a graph
    
    Args:
        graph: adjacency list representation
        start: starting node
        visited: set of visited nodes
    
    Returns:
        list of nodes in DFS order
    """
    if visited is None:
        visited = set()
    
    visited.add(start)
    result = [start]
    
    # Visit all neighbors
    for neighbor in graph[start]:
        if neighbor not in visited:
            # extend works like append but for list with list
            # e.g. [A, B] append [C, D] = [A, B, [C, D]]
            #      [A, B] extend [C, D] = [A, B, C, D]
            result.extend(dfs_recursive(graph, neighbor, visited))
    
    return result

# Test DFS
sample_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}

print("DFS Traversal starting from 'A':")
print(dfs_recursive(sample_graph, 'A'))
```

```python
# DFS Implementation (Iterative using Stack)

def dfs_iterative(graph, start):
    """
    Perform Depth-First Search using stack
    """
    visited = set()
    stack = [start]
    result = []
    
    while stack:
        node = stack.pop()
        
        if node not in visited:
            visited.add(node)
            result.append(node)
            
            # Add neighbors to stack (in reverse order to maintain left-to-right traversal)
            for neighbor in reversed(graph[node]):
                if neighbor not in visited:
                    stack.append(neighbor)
    
    return result

print("DFS Iterative starting from 'A':")
print(dfs_iterative(sample_graph, 'A'))
```

#### 3. Breadth-First Search (BFS)

**Concept**: Explore all neighbors at the present depth before moving to nodes at the next depth level

**Implementation**: Using queue

```python
# BFS Implementation using Queue - Similar to our Level-Order Traversal!
from collections import deque

def bfs(graph, start):
    """
    Perform Breadth-First Search on a graph
    
    Args:
        graph: adjacency list representation
        start: starting node
    
    Returns:
        list of nodes in BFS order
    """
    visited = set()
    queue = deque([start])
    visited.add(start)
    result = []
    
    while queue:
        node = queue.popleft()
        result.append(node)
        
        # Add all unvisited neighbors to queue
        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append(neighbor)
    
    return result

sample_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E']
}


print("BFS Traversal starting from 'A':")
print(bfs(sample_graph, 'A'))
```
