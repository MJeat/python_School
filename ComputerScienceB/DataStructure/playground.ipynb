```python
import networkx as nx
import matplotlib.pyplot as plt
import numpy as np

# Adjacency list
adj_list = {
    'A': ['B'],
    'B': ['C'],
    'C': ['E'],
    'D': ['B'],
    'E': ['D', 'F'],
    'F': []
}

# Create directed graph from adjacency list
G_list = nx.DiGraph(adj_list)

# Plot graph from adjacency list
plt.figure(figsize=(8,4))
plt.subplot(121)
nx.draw(G_list, with_labels=True, node_size=1500, node_color='lightblue', arrowsize=20)
plt.title("Graph from adjacency list")

# Adjacency matrix
nodes = ['A', 'B', 'C', 'D', 'E', 'F']
adj_matrix = np.array([
    [0, 1, 0, 0, 0, 0],  # A
    [0, 0, 1, 0, 0, 0],  # B
    [0, 0, 0, 0, 1, 0],  # C
    [0, 1, 0, 0, 0, 0],  # D
    [0, 0, 0, 1, 0, 1],  # E
    [0, 0, 0, 0, 0, 0]   # F
])

# Create graph from adjacency matrix
G_matrix = nx.DiGraph(adj_matrix)
mapping = {i: node for i, node in enumerate(nodes)}
G_matrix = nx.relabel_nodes(G_matrix, mapping)

# Plot graph from adjacency matrix
plt.subplot(122)
nx.draw(G_matrix, with_labels=True, node_size=1500, node_color='lightgreen', arrowsize=20)
plt.title("Graph from adjacency matrix")

plt.show()
```
