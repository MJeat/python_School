#### 2. Binary Tree

```python
# Node class for Binary Tree
class Node:
    def __init__(self, data):
        self.data = data
        self.left = None
        self.right = None

# Binary Tree implementation
class BinaryTree:
    def __init__(self):
        self.root = None
    
    # Insert a new value (level-order insertion)
    def insert(self, data):
        new_node = Node(data)
        
        if self.root is None:
            self.root = new_node
            return
        
        # Use a queue for level-order insertion
        queue = [self.root]
        
        while queue:
            current = queue.pop(0)
            
            # Insert at the first available position
            if current.left is None:
                current.left = new_node
                return
            else:
                queue.append(current.left)
            
            if current.right is None:
                current.right = new_node
                return
            else:
                queue.append(current.right)
    
    # Search for a value (level-order search)
    def search(self, data):
        if self.root is None:
            return False
        
        queue = [self.root]
        
        while queue:
            current = queue.pop(0)
            
            if current.data == data:
                return True
            
            if current.left:
                queue.append(current.left)
            if current.right:
                queue.append(current.right)
        
        return False
    
    # In-order traversal: Left → Root → Right
    def inorder_traversal(self):
        result = []
        self._inorder_recursive(self.root, result)
        return result
    
    def _inorder_recursive(self, node, result):
        if node:
            self._inorder_recursive(node.left, result)
            result.append(node.data)
            self._inorder_recursive(node.right, result)
    
    # Pre-order traversal: Root → Left → Right
    def preorder_traversal(self):
        result = []
        self._preorder_recursive(self.root, result)
        return result
    
    def _preorder_recursive(self, node, result):
        if node:
            result.append(node.data)
            self._preorder_recursive(node.left, result)
            self._preorder_recursive(node.right, result)
    
    # Post-order traversal: Left → Right → Root
    def postorder_traversal(self):
        result = []
        self._postorder_recursive(self.root, result)
        return result
    
    def _postorder_recursive(self, node, result):
        if node:
            self._postorder_recursive(node.left, result)
            self._postorder_recursive(node.right, result)
            result.append(node.data)
    

# Example usage of Binary Tree
bt = BinaryTree()

# Insert values: 50, 30, 70, 20, 40, 60, 80
values = [50, 30, 70, 20, 40, 60, 80]
print("Inserting values:", values)
for value in values:
    bt.insert(value)

print("\nTree structure (level-order insertion):")
print("       50")
print("      /  \\")
print("    30    70")
print("   /  \\   /  \\")
print("  20  40 60  80")

# Search operations
print("\nSearch Results:")
print("Is 40 in the tree?", bt.search(40))
print("Is 100 in the tree?", bt.search(100))

# Traversal operations
print("\nTraversal Results:")
print("In-order (L→R→R):  ", bt.inorder_traversal())
print("Pre-order (R→L→R): ", bt.preorder_traversal())
print("Post-order (L→R→R):", bt.postorder_traversal())
```

#### 3. Heaps (Priority Queue)

```python
import heapq

# Create an empty heap
heap = []

# Push elements into the heap. By default, heap is implemented as min-heap.
print("Pushing elements: 5, 2, 8, 1, 9, 3")
heapq.heappush(heap, 5)
heapq.heappush(heap, 2)
heapq.heappush(heap, 8)
heapq.heappush(heap, 1)
heapq.heappush(heap, 9)
heapq.heappush(heap, 3)

print("Heap:", heap)
print("Smallest element (peek):", heap[0])

# Pop smallest element
smallest = heapq.heappop(heap)
print("Popped smallest element:", smallest)
print("Heap after pop:", heap)

# Convert a list to a heap
numbers = [7, 3, 9, 1, 5, 4]
print("\nOriginal list:", numbers)
heapq.heapify(numbers)
print("After heapify:", numbers)

# Find n largest elements
data = [5, 2, 8, 1, 9, 3, 7, 4, 6]
print("\nData:", data)
print("3 largest elements:", heapq.nlargest(3, data))
print("3 smallest elements:", heapq.nsmallest(3, data))
```

##### Heaps - Max Heap Implementation

```python
# Python's heapq only supports min heap
# To create a max heap, we negate the values

max_heap = []

# Push elements (negate for max heap)
print("Pushing elements for max heap: 5, 2, 8, 1, 9, 3")
for value in [5, 2, 8, 1, 9, 3]:
    heapq.heappush(max_heap, -value)

print("Max heap (stored as negatives):", max_heap)

# Pop largest element (negate when retrieving)
largest = -heapq.heappop(max_heap)
print("Popped largest element:", largest)
print("Current largest element:", -max_heap[0])
```

We import the deque for this code. deque is used as a queue to keep track of nodes to visit. It is great for this task of popping or adding something. The variable "q" makes a new queue and puts the root node into it to start. The queue "q" holds nodes we still need to visit. Starting with the root means “we will visit the root first.” Then, we loop through the length of the queue. 

While the queue is not empty, it removes the front node, prints it, and adds its left and right children (if they exist) to the end of the queue. This ensures that nodes are visited row by row, from left to right. 


This is a level-order or breadth-first traversal.
