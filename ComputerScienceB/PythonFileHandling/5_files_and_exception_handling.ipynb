#### Exception Handling

```python
file_path = "exception_handling.txt"
try:
    file = open(file_path, "r")
    print(file.read())
    file.close()
except Exception as e:
    print("There is an error occurred.")
    print(e)

print("The program still continues!")
```

#### Retrieving Data from the Web

```python
from urllib import request

url = "https://www.gutenberg.org/cache/epub/1513/pg1513.txt"    # Romeo and Juliet from Project Gutenberg
response = request.urlopen(url)
raw = response.read().decode('utf8')
print(raw[:500])    # print first 500 characters
```

```python
from urllib import request

url = "https://www.google.com/search?q=python"
response = request.urlopen(url)
raw = response.read().decode('utf8')
print(raw[:500])    # print first 500 characters

# There are external library for dealing with html files
```

#### Binary File

```python
import pickle

# Writing to binary file
outfile = open("binary_file.dat", "wb")
pickle.dump(45, outfile)
pickle.dump(56.6, outfile)
pickle.dump("Programming is fun", outfile)
pickle.dump([1, 2, 3, 4], outfile)
outfile.close() # Close the output file
```

```python
# Open binary file for reading
infile = open("binary_file.dat", "rb")
print(pickle.load(infile))
print(pickle.load(infile))
print(pickle.load(infile))
print(pickle.load(infile))
infile.close() # Close the input file
```

```python
# Can also use the with open() syntax
with open("binary_file_2.dat", "wb") as outfile:
    pickle.dump(15, outfile)
    pickle.dump("Programming is fun", outfile)
    pickle.dump([1, 2, 3, 4], outfile)
    pickle.dump("I love Python", outfile)

with open("binary_file_2.dat", "rb") as infile:
    print(pickle.load(infile))
    print(pickle.load(infile))
    print(pickle.load(infile))
    print(pickle.load(infile))
```
