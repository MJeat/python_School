```python

def reverseDisplay(value):
    """
    Print the characters of `value` in reverse order using recursion.
    This function prints characters directly (doesn't return the reversed string).
    """
    if value == "":
        # base case: nothing left to print
        return
    # print the last character, without newline
    print(value[-1], end="")
    # recurse on the remaining string (all except the last char)
    reverseDisplay(value[:-1])



s = input("Enter a string: ")
reverseDisplay(s)


```

```python
def count(s, a):

    if s == "":
        # base case: empty string → no occurrences
        return 0
    
    # check the first character
    if s[0] == a:
        return 1 + count(s[1:], a)  # found one, check the rest
    else:
        return count(s[1:], a)      # not a match, check the rest



s = input("Enter a string: ")
a = input("Enter the character to count: ")
print(f"'{a}' occurs {count(s, a)} times in '{s}'.")
```

```python
def decimalToBinary(value):
    # base case: 0 or 1
    if value < 2:
        return str(value)

    return decimalToBinary(value // 2) + str(value % 2)

n = int(input("Enter a number: "))
print(decimalToBinary(n))
```
