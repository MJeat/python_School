1. Reverse String

```python
# Option 1: The function will do the printing
def reverseDisplay(val):
    # base case - since we're printing, we want to print new line as well
    if len(val) == 0:
        print()
    else:
        print(val[-1], end="") # print last character
        reverseDisplay(val[:-1]) # recursively work on remaining characters at the front

# Option 2: The function will return the reversed string
def reverseDisplay2(val):
    # base case - since we're only returning the string, we will let users handle the print line
    if len(val) == 1:
        return val  # once reach only 1 char, start returning it back

    return val[-1] + reverseDisplay2(val[:-1])  # last char + recursively on remianing char

test_str = input("Input your string to test: ")
reverseDisplay(test_str)
print(reverseDisplay2(test_str))
```

2. Count occurences in a String

```python
def count(s, a):
    # count a and store it in count_curr
    count_curr = 0
    if s[0].lower() == a.lower():
        count_curr = 1
    
    # base case
    if len(s) == 1:
        return count_curr
        
    return count_curr + count(s[1:], a) # count a in the first char + recursively count in the remaining string

test_str = input("Input your string to test: ")
test_char = input("Input a single character to check: ")
print(count(test_str, test_char))
```

3. Decimal to Binary

```python
def decimalToBinary(value):
    # base case: 1 // 2 = 1
    if value == 1:
        return "1"

    return str(decimalToBinary(value//2)) + str(value % 2)
    # e.g. 10 -> 1010
    # starts with 10 % 2 = 0, this will have to go at the end of the string
    # recursively do it on 10//2 (5) and then append it to the result of (10 % 2)

test_str = int(input("Input your string to test: "))
print(decimalToBinary(test_str))
```
