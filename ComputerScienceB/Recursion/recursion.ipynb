### 1. Factorial

```python
# Using loop
n = 5
factorial = 1
for i in range(1, n+1):
    factorial *= i

print(factorial)
```

```python
# Using recursion
def calc_factorial(n):
    # base case
    if n == 0 or n == 1:
        return 1
    
    return n * calc_factorial(n-1) 

factorial_rec = calc_factorial(n)
print(factorial_rec)
```

```python
def calc_power(x, n):
    # Write your code here
    # base case:
    if n==1:
        return x
    return x*calc_power(x,n-1) # Start from the recursive state first

# Recursive function to calculate power of a number
x = 2
n = 5
# 2^5 = 32
print(calc_power(x, n))
```

### 2. Binary Search

```python
# Without recursion
num = list(range(1, 50))

def binary_search(search_num, lst):
    lower = 0
    upper = len(lst) - 1

    while lower <= upper: 
        mid = (lower + upper) // 2
        
        if lst[mid] == search_num:
            return mid  # Return index when found
        elif search_num < lst[mid]:
            upper = mid - 1     # Search upper half
        else:
            lower = mid + 1     # Search lower half
    
    return -1  # Element not found

print(binary_search(5, num))
```

```python
# Using recursion
def binary_search_recursive(search_num, lst, lower=0, upper=None):
    # Initialize upper bound on first call
    if upper is None:
        upper = len(lst) - 1
    
    # Base case: element not found
    if lower > upper:
        return -1
    
    # Calculate our mid
    mid = (lower + upper) // 2
    
    # Base case: element found
    if lst[mid] == search_num:
        return mid
    elif search_num < lst[mid]:     # Recursive case: search lower half
        return binary_search_recursive(search_num, lst, lower, mid - 1)
    else:                           # Recursive case: search upper half
        return binary_search_recursive(search_num, lst, mid + 1, upper)
```

```python
# Using recursive helper
def binary_search_recursive_helper(search_num, lst):
    lower = 0
    upper = len(lst) - 1
    return binary_search_recursive_new(search_num, lst, lower, upper)

def binary_search_recursive_new(search_num, lst, lower, upper):
    if lower > upper:
        return -1
    
    mid = (lower + upper) // 2
    # Base case: element found
    if lst[mid] == search_num:
        return mid
    elif search_num < lst[mid]:
        return binary_search_recursive(search_num, lst, lower, mid - 1)
    else:
        return binary_search_recursive(search_num, lst, mid + 1, upper)
```
