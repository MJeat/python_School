Here are other common problems that are solved using Recursion:  Triangular numbers, Fibonacci, Anagrams, palindromes, Towers of Hanoi, and Mergesort


* Triangular numbers – adding numbers from 1 up to n.
* Fibonacci sequence – finding each number as the sum of the two before it.
* Anagrams – generating all possible word arrangements.
* Palindromes – checking if a word reads the same backward.
* Towers of Hanoi – moving disks between pegs following rules.
* Mergesort – sorting a list by dividing it into smaller parts

```python
# Triangular numbers – adding numbers from 1 up to n.
def triangle(n):
    if n == 1:
        return 1
    return n + triangle(n - 1)

print(triangle(5))  # 1 + 2 + 3 + 4 + 5 = 15

```

```python
# Fibonacci sequence – finding each number as the sum of the two before it. 
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

print(fib(6))  # 8
```

```python
# Anagrams – generating all possible word arrangements.
def anagram(word):
    if len(word) == 1:
        return [word]
    result = []
    for i, ch in enumerate(word):
        for sub in anagram(word[:i] + word[i+1:]):
            result.append(ch + sub)
    return result

print(anagram("cat"))
```

```python
# Palindromes – checking if a word reads the same backward.
def is_palindrome(word):
    if len(word) <= 1:
        return True
    if word[0] != word[-1]:
        return False
    return is_palindrome(word[1:-1])

print(is_palindrome("madam"))  # True
```

```python
# Towers of Hanoi – moving disks between pegs following rules.
def hanoi(n, source, target, helper):
    if n == 1:
        print(f"Move disk 1 from {source} to {target}")
        return
    hanoi(n - 1, source, helper, target)
    print(f"Move disk {n} from {source} to {target}")
    hanoi(n - 1, helper, target, source)

hanoi(3, 'A', 'C', 'B')
```

```python
# Mergesort – sorting a list by dividing it into smaller parts 
def merge_sort(arr):
    if len(arr) <= 1:
        return arr
    mid = len(arr) // 2
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    while left and right:
        if left[0] < right[0]:
            result.append(left.pop(0))
        else:
            result.append(right.pop(0))
    return result + left + right

print(merge_sort([5, 2, 8, 1, 3]))
```
