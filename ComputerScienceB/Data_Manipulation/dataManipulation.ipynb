2D List

```python
# 2D list

rows = int(input("Enter the number of rows: "))
cols = int(input("Enter the number of rows: "))

my_2d_list = []

for i in range(rows):
    # creating empty row
    my_2d_list.append([])
    for j in range(cols):
        my_2d_list[i].append(j) # fill each col to the row

print(my_2d_list)
```
