# OOP

```python
# Inheritance


class SchoolMember:
    def __init__(self, name, id):
        self.name = name
        self.id = id
    def structure(self):
        print(f"Name: {self.name}")
        print(f"ID: {self.id}")

class Teacher(SchoolMember):
    def __init__(self, name, id, specialization):
        super().__init__(name, id)
        self.specialization = specialization
    def teach(self):
        print(f"Teach: {self.specialization}")

class Student(SchoolMember):
    def __init__(self, name, id, major):
        super().__init__(name, id)
        self.major = major
    def subject(self):
        print(f"Teach: {self.major}")


teacher = Teacher("Bob", 23, "CS")
teacher.structure()
teacher.teach()

student = Student("Amy", 12, "law")
student.structure()
student.subject()
        
```

```python
# Polymorphism - We are reusing the sound() function from different classes, including parent and child classes
class Animal:
    def sound(self):
        return "Bob"
class Dog(Animal):
    def sound(self):
        return "Woof"
    
class Cat(Animal):
    def sound(self):
        return "Meow"
    
animals = [Dog(), Cat(), Animal()]
for animal in animals:
    print(animal.sound())
```

```python
# Encapsulation

class Information:
    def __init__(self, name, id, age, credentials):
        self.name = name
        self.id = id
        self._age = age
        self.__credentials = credentials
    def showDetails(self):
        print(f"Name: {self.name}")
        print(f"ID: {self.id}")
        print(f"Age: {self._age}")
        print(f"Password: {self.__credentials}")

class Member(Information):
    def __init__(self, name, id, age, credentials, subject):
        super().__init__(name, id, age, credentials)
        self.subject = subject
    def display(self):
        print(f"Subject: {self.subject}")

teacher = Member("Anna", 23, 53, "123456", "CS")
teacher.display()
print(teacher.name)
# print(teacher.__credentials) # This will show error because it cannot print since it's private
print(teacher._age) # This will show because it's just protected from data leak or data manipulation
```

```python
# Abstraction - This hides the code from the user to avoid them exploiting.

from abc import ABC, abstractmethod
class Animal(ABC): # Abstract base class
    @abstractmethod
    def make_sound(self):
        pass # Must be implemented by subclasses
    @abstractmethod
    def move(self):
        pass
    # You cannot instantiate Animal directly
    # animal = Animal() # This would raise TypeError
class Dog(Animal):
    def make_sound(self):
        return "Woof!"
    def move(self):
        return "Running on four legs"
class Bird(Animal):
    def make_sound(self):
        return "Chirp!"
    def move(self):
        return "Flying with wings"
husky = Dog()
print(husky.make_sound())
print(husky.move())

```

# Recursion

Tips:

In recursion, if the range is from 1 until 100, we work from 100 and stops at 1 or at the base component. So, make sure the code starts printing from big to small, and then dominos back, finishing the overall operation/calculation

```python
# Practice 1 (Not recursion): Factorial

# I want to input the number 5 and I want the factorial of 5
print("===Iterative Method===")
n =5
factorial_value = 1
for i in range(1, n+1):
    factorial_value *= i
print(factorial_value)


# Recursion Method
print("===Recursion Method===")
def factorial(n):
    if n == 1:
        return 1
    return n*factorial(n-1)
print(factorial(5))
```

```python
# Practice 2: Power
print("===Recursion Method===")
# You need to find the base component. This is a part where to want the recursion to reach a stop and return the value back to previous functions.

def calc_power(x,y):
    if y == 1:
        return x
    return x * calc_power(x, y-1)

print(calc_power(2,7))
```

```python
# Practice 3: Sum of Digits
# Write a recursive function called sum_of_digits(n) that returns the sum of all digits in a positive integer.
# sum_of_digits(1234)  # should return 10 because 1+2+3+4 = 10
# Hint:
    # Each time, extract the last digit using % 10.
    # Remove the last digit using // 10.
    # Use recursion until n becomes 0.
"""
n % 10: This extracts the last digit of the number. For example, 12345 % 10 gives 5.
n // 10: This performs integer division, effectively removing the last digit from the number. For example, 12345 // 10 gives 1234.
"""

def sumOfDigits(n):
    if n == 0:
        return 0
    return (n%10) + sumOfDigits(n//10)
sumOfDigits(1234)
```

```python
# Practice 4: Count Digits
# Write a function called count_digits(n) that takes a positive integer n and returns how many digits it has — using recursion

def count_digits(n):

    # Handle negative numbers
    if n <0:
        n = abs(n)
    # Base case
    if n < 10:
        return 1
    return 1 + count_digits(n//10)


print(count_digits(1234), "Digits")
print(count_digits(-12), "Digits")
print(count_digits(0), "Digits")
```

```python
# Practice 5: Reverse String
# Write a function called reverse_string(s) that takes a string s and returns the string reversed, but you cannot use slicing ([::-1]) or loops. You must use recursion only.

def reverse_string(s):

    if s == "":
        return s

    return s[-1] + reverse_string(s[:-1]) 


print(reverse_string("olleh"))

# this is very similar to sumOfDigit recursion
# s[-1] means to take the last character. Since its at the front, that last character will comes to the front
# s[:-1] means to not take the last character. Means to take any strings that are not the last character and continue the recursion
```
