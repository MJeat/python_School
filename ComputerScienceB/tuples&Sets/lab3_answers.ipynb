#### Exercise 2
You are given a list of student and their score. 
- Your task is to create a list of tuples where each tuple contains the student's name and their letter grade.
- After that output the tuples, and the average score.
- Then, from the list of tuples, output the grade that has the most students, all the students with grade A, and all the students with grade F.

Grading Scale:

- A: 90-100
- B: 80-89
- C: 70-79
- D: 60-69
- F: Below 60

```python
# 20 students with 20 scores in order
students = ["Sophea", "David", "Pisey", "Emma", "Sokha",
            "James", "Channary", "Oliver", "Sreypov", "Liam",
            "Chenda", "Davy", "Pisal", "Isabella", "Virak",
            "Mason", "Bopha", "Eva", "Puthea", "Noah"]
    
scores = [85, 92, 67, 88, 79,
            95, 80, 82, 91, 58,
            74, 89, 93, 71, 86,
            64, 78, 87, 90, 52]

# Write your code here
def get_grade(score):
    grade = "Invalid"
    if score >= 90 and score <= 100:
        grade = 'A'
    elif score >= 80:
        grade = 'B'
    elif score >= 70:
        grade = 'C'
    elif score >= 60:
        grade = 'D'
    elif score >= 0:
        grade = 'F'
    else:
        grade = 'Invalid'
    return grade

def most_students_grade(tuples_lst):
    # Find grade with most students
    # Create a dictionary for grade count
    grade_count = {}
    for item in tuples_lst:
        if item[1] in grade_count.keys():
            grade_count[item[1]] += 1
        else:
            grade_count[item[1]] = 1
    return grade_count

def print_student_grade(tuples_lst, grade):
    print(f"\nStudents with grade {grade}:")
    for stu in tuples_lst:
        if stu[1] == grade:
            print(stu[0])


stu_tuples_lst = []
for i in range(len(students)):
    # Add tuples (student, grade) to list
    tuple_to_add = (students[i], get_grade(scores[i]))
    stu_tuples_lst.append(tuple_to_add)

print(stu_tuples_lst)
print(f"\nAverage score = {(sum(scores) / len(students)):.2f}")

# Find grade with most students
grade_count = most_students_grade(stu_tuples_lst)

# Get key with the maximum values
max_key = max(grade_count, key=grade_count.get)
# Get the maximum values
max_val = max(grade_count.values())
print(f"\n{max_val} out of {len(students)} students have grade {max_key}.")

print_student_grade(stu_tuples_lst, 'A')
print_student_grade(stu_tuples_lst, 'F')

# I'm using function here for cleaner code, but even if you don't use function, it's fine too.
```
