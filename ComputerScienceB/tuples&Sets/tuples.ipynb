This is the testing tuple in Python. Tuple is a compound type. Meaning it stores a lot of data and can be different data types.

```python
tuple1 = ("green", "blue", "black")
print(tuple1)

tuple2 = tuple([3,2,6,1,4,8]) #converting list to tuple
print(tuple2)

```

```python
print("length is: ", len(tuple2)) #length of tuple2
print("max is: ", max(tuple2)) #max value in tuple2
print("min is: ", min(tuple2)) #min value in tuple2
print("sum is: ", sum(tuple2)) #sum of all values in tuple2

print("The first element is: ", tuple2[0]) #accessing first element
print("The last element is: ", tuple2[-1]) #accessing last element
```
