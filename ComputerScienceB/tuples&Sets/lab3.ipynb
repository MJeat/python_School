#### Exercise 2
You are given a list of student and their score. 
- Your task is to create a list of tuples where each tuple contains the student's name and their letter grade.
- After that output the tuples, and the average score.
- Then, from the list of tuples, output the grade that has the most students, all the students with grade A, and all the students with grade F.

Grading Scale:

- A: 90-100
- B: 80-89
- C: 70-79
- D: 60-69
- F: Below 60

```python
# 20 students with 20 scores in order
students = ["Sophea", "David", "Pisey", "Emma", "Sokha",
            "James", "Channary", "Oliver", "Sreypov", "Liam",
            "Chenda", "Davy", "Pisal", "Isabella", "Virak",
            "Mason", "Bopha", "Eva", "Puthea", "Noah"]
    
scores = [85, 92, 67, 88, 79,
            95, 80, 82, 91, 58,
            74, 89, 93, 71, 86,
            64, 78, 87, 90, 52]

# Write your code here

# Convert score to grade
grading=[]
for i in scores:
    if 90<=i<=100:
        grade = 'A'
    elif 80<=i<=89:
        grade='B'
    elif 70<=i<=79:
        grade='C'
    elif 60<=i<=69:
        grade='D'
    elif i < 60:
        grade='F'
    else:
        print("Error Here")
    grading.append(grade)
print()

# Prinint names with grades
name_with_score = []
for i in range(len(students)):
    name_with_score.append((students[i], grading[i]))
print(name_with_score)

# Number of students
scores_len=len(scores)

# Average score
avg_score = 0
for i in scores:
    avg_score+=i/scores_len
print(f"Average Score: {avg_score:.2f}")
print()

# Student that got b
got_b=0
for i in grading:
    if i=='B':
        got_b+=1
print(f"{got_b} out of {scores_len} got grade B")
print()

# Students that got A
print("Students with grade A: ")
got_A=''
for i,j in name_with_score:
    if j =="A":
        print(i)
print()

# Students who fail
print("Students who failed the course: ")
for i,j in name_with_score:
    if j =="F":
        print(i)
print()

# The grade that has the most student
count=max(set(grading), key=grading.count)
print("A Grade that has the most students: "+count)
```
