A set is an unordered collection with no duplicate elements.
- Mutable, but elements must be immutable.
E.g. Sets cannot contain a list (mutable).
- Automatically removes duplicates.
- Does not support indexing
- Can mix different data types
- set() is the empty constructor
- Also uses {} curly braces



```python
s1 = set() #creating an empty set
s2 = {1,3,5,7,9} #creating a set with elements
s3 = set((1,3,5,7,9)) #creating a set from a tuple

# Creating a set from a list
s4 = set([x*2 for x in range(1,10)])
s5 = set("aaaabdc") #automatically removes duplicates
s6 = {1,1,2,3,3}

print(s1)
print(s2)
print(s3)
print(s4)
print(s5)
print(s6)
```

```python
s1 = {1,2,4}
s2 = {1,3,2,6,5,4}
print(s1.issubset(s2)) #check if s1 is a subset of s2
print(s1.issuperset(s2)) #check if s1 is a superset of s2
print(s2.issuperset(s1)) #check if s2 is a superset of s1
```

This is about the set operations. It has Union, Intersaction, Difference, and Symmetric Difference.

- Union – join two sets together
(automatically remove duplicates).
- Intersection – keep only the same items
of both sets.
- Difference – setA.difference(setB) will
keep only what in setA that setB doesn’t
have.
- Symmetric Difference – keep all the
differences on both sets

```python
# Union
print("Union")
s1 = {1,2,4}
s2 = {1,3,2,6,5,4}
print(s1.union(s2)) #or s1 | s2
print(s1 | s2)

# Intersection
print("Intersection")
print(s1.intersection(s2)) #or s1 & s2
print(s1 & s2)

# Difference
print("Difference")
print(s1.difference(s2)) #or s1 - s2
print(s1 - s2)

# Symmetric Difference
print("Symmetric Difference")
print(s1.symmetric_difference(s2)) #or s1 ^ s2
print(s1 ^ s2)
```
