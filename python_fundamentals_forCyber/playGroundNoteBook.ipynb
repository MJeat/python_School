```python
# Practice 1
# Sending requests to websites to gain the responses
import requests
from bs4 import BeautifulSoup

url = "http://quotes.toscrape.com"
response = requests.get(url)

print(response.status_code)
print(response.url) # after redirection, what's the final destination?

# Step 2: Parse the page content with BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')
print(type(soup))

# Step 3: Extract all quote text and authors
quotes = soup.find_all('div', class_='quote')
print(quotes[1])

# Step 4: Print each quote and its author
for quote in quotes:
    text = quote.find('span', class_='text').get_text()
    author = quote.find('small', class_='author').get_text()
    print(f"{text} — {author}")
```

```python
# Practice 2
# Scrape from https://quotes.toscrape.com/


url2 = "https://quotes.toscrape.com/"
pageToScrape2 = requests.get(url2)
soup = BeautifulSoup(pageToScrape2.text, "html.parser")

authorQuotes = soup.find_all("span", attrs={"class":"text"})
authors = soup.find_all("small", class_="author")

for i, j in zip(authorQuotes, authors):
    print(f"{i.text} - {j.text}")

```

# IV. Cryptography

```python
import hashlib
import base64

# Part 1: Hash messages
print("=== PART 1: Hashing ===")
messages = ["password123", "CyberSecurity", "hello world", "123456"]

for msg in messages:
    sha256_hash = hashlib.sha256(msg.encode()).hexdigest()
    md5_hash = hashlib.md5(msg.encode()).hexdigest()
    print(f"Message: {msg}")
    print(f"SHA256:  {sha256_hash}")
    print(f"MD5:     {md5_hash}\n")

```

```python
# Part 2: Base64 Encoding
print("=== PART 2: Base64 Encoding ===")
for msg in messages:
    encoded = base64.b64encode(msg.encode()).decode()
    decoded = base64.b64decode(encoded).decode()
    print(f"Original: {msg}")
    print(f"Base64:   {encoded}")
    print(f"Decoded:  {decoded}\n")

```

```python
# Part 3: Combine Hashing + Base64
print("=== PART 3: Hashing + Base64 ===")
for msg in messages:
    # Get SHA256 hash as bytes
    sha256_bytes = hashlib.sha256(msg.encode()).digest()
    # Convert to hex and base64
    sha256_hex = sha256_bytes.hex()
    sha256_b64 = base64.b64encode(sha256_bytes).decode()
    
    print(f"Message: {msg}")
    print(f"SHA256 (hex):  {sha256_hex}")
    print(f"SHA256 (b64):  {sha256_b64}\n")
```
