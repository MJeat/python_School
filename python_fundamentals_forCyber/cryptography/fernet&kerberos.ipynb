# Fernet & Kerberos

```python
import sys

print('Python' , sys.version.splitlines()[0])

try:
    from cryptography.fernet import Fernet, InvalidToken
    from cryptography.hazmat.primitives.kdf.pbkdf2 import PBKDF2HMAC
    from cryptography.hazmat.primitives import hashes
    from cryptography.hazmat.backends import default_backend
    import base64, os
    print('Cryptography Imported OK')

except Exception as e:
    print('Import Issue: ',e)
```

```python
# Generate a new key
key = Fernet.generate_key()
print('Generated key (Base64): ',key)

# Create a Fernet object
f = Fernet(key)

# Encrypt
# msg = 'សួស្ដីបង' # If the character is not in ASCII table, don't need to put b - for byte
msg = b'Hello guys'
token = f.encrypt(msg)
print('Encrypted Token: ',token)
```

```python
# Decrypt the message
plain = f.decrypt(token.decode("utf-8"))
print('Token (String): ', token)

original = f.decrypt(token).decode("utf-8")
print("Original: ",original)
```

# Working With Files

```python

# Write the file
file = 'Sample.txt' 
with open(file, 'wb') as fsrc: # file source
    fsrc.write(b"Important File Contents. Hello!")

# Encrypt the file -> .enc  - This is encrypted file
with open(file, 'rb') as fsrc:
    data = fsrc.read()
    enc = f.encrypt(data)


with open(file + '.enc', 'wb') as fout: # file output
    fout.write(enc)
print('Wrote encrypted file: ', file+'.enc')


# Decrypt 
with open(file + '.enc','rb') as fin: # file input
    tok = fin.read()
plain = f.decrypt(tok)
print("Decrypted File Content: ",plain)
```
