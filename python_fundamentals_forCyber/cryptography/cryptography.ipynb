# I. Using hashlib

```python
import hashlib


messages = [b"hello world",b"hello worlD",b"hello world!",] # "b" here is byte

for eachm in messages:
    eachh = hashlib.sha256(eachm).hexdigest()
    print(type(eachh))
    
    print(eachm.decode('utf-8', errors='replace'), ' : ', eachh)

# errors='replace' here means that if the strings in messages are not in the ascii table, it will replace with something else.
# This is to prevent error or broken codes
```

# III. Crypto

```python
import os
import hashlib
import binascii

Pepper = os.getenv('APP_PEPPER', 'Pappeer_used_for_cybr353_only')
KDF_NAME = 'sha256'
KDF_Interations = 200000
KDF_DKLEN = 32



message = [b'hello world ', b'hello World', b'hello world!', b'Hello world']
for each in message:
    fast_hash = hashlib.sha256(each).hexdigest()
    salt = os.urandom(16)
    salt_hex = binascii.hexlify(salt).decode()
    peppered = each + Pepper.encode()

    derived_key = hashlib.pbkdf2_hmac(
        KDF_NAME,    
        peppered,
        salt,
        KDF_Interations,
        dklen = KDF_DKLEN)
    derived_hex = binascii.hexlify(derived_key).decode()
    print(each.decode('utf-8', errors = 'replace'), ": ", fast_hash, " ", salt_hex, " ", derived_hex)
```

# IV. Base64

```python
import base64

text = "welcome everyone"
encoded = base64.b64encode(text.encode())
decoded = base64.b64decode(encoded).decode()

print("Original: ", text)
print("Base64: ", encoded)
print("Decoded: ", decoded)
```
