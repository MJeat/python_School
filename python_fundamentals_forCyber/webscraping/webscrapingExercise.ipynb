This part is to initialize the libraries and the main header of the website

```python
import requests
from bs4 import BeautifulSoup

pageToScrape = requests.get("https://thehackernews.com/2025/09/evilai-malware-masquerades-as-ai-tools.html")
soup = BeautifulSoup(pageToScrape.text, "html.parser")

title = soup.find_all("h1", attrs={"class":"story-title"})
p_author = soup.find("span", class_="p-author")

for i in title:
    print(f"Story Title: \n{i.text}\n")

if p_author: # Use this if what you want is inside another container above
    # Find all <span class="author"> inside it
    author_spans = p_author.find_all("span", class_="author")
    
    if len(author_spans) >= 2:
        date = author_spans[0].get_text(strip=True)
        author = author_spans[1].get_text(strip=True)
        
        print("Date:", date)
        print("Author:", author)
    else:
        print("Not enough author spans found")
else:
    print("p-author span not found")
```

This part is to get the contents

```python
article = soup.find("div", class_="articlebody clear cf")

if article:
    # Find all <p> tags inside it
    paragraphs = article.find_all("p")
    
    for p in paragraphs:
        text = p.get_text(strip=True)
        if text:  # Only print non-empty paragraphs
            print(text)
else:
    print("Article container not found.")
```
