This is a web scraping practice with notebook

```python
import requests
from bs4 import BeautifulSoup

pageToScrape = requests.get("https://quotes.toscrape.com/")
soup = BeautifulSoup(pageToScrape.text, "html.parser")

authorQuotes = soup.find_all("span", attrs={"class":"text"})
authors = soup.find_all("small", attrs={"class":"author"})

# for i in authorQuotes:
#     print(i.text)
# print("============")

# for j in authors:
#     print(j.text)
# print("============")

for i,j in zip(authorQuotes,authors): # Using zip() to pair two different variables
    print(f"{i.text} - {j.text}")

```

Getting the tags only

```python
bigTags = soup.find_all("span", attrs={"class":"tag-item"})
for z in bigTags:
    print(z.text, end='') # Using end='' to decrease the space between above and below words
```
