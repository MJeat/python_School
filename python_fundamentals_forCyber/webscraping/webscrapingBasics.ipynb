To scrap data from websites

```python
import requests
from bs4 import BeautifulSoup
 

# Step 1: Send a request to the website
url = "http://quotes.toscrape.com"
response = requests.get(url)
print(response)

print(response.status_code)     # 200
print(response.ok)              # True
print(response.headers['Content-Type'])  # 'text/html; charset=UTF-8'
print(response.url)             # Final URL after redirects
print(response.text[:500])      # First 500 characters of HTML
```

```python
# Step 2: Parse the page content with BeautifulSoup
soup = BeautifulSoup(response.text, 'html.parser')
print(type(soup))
```

```python
# Step 3: Extract all quote text and authors
quotes = soup.find_all('div', class_='quote')
print(quotes[1])
```

```python
# Step 4: Print each quote and its author
for quote in quotes:
    text = quote.find('span', class_='text').get_text()
    author = quote.find('small', class_='author').get_text()
    print(f"{text} — {author}")
```
