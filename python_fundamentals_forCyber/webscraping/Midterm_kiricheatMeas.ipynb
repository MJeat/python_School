Fetch the page and check status:
* Use requests to GET the URL.

* Print the HTTP status code.

* If status is not 200, exit with an error message.

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.projecthoneypot.org/list_of_ips.php?by=3&ctry=KH"
response = requests.get(url)
status = response.status_code
if status != 200:
    print("Error")
print(status)
```

Parse with BeautifulSoup

* Use BeautifulSoup (with the "html.parser" or "lxml" parser) to parse the HTML
contents.

```python
soup = BeautifulSoup(response.text, "html.parser")
print(soup)
```

Inspect headers
* Print out all response headers from requests (i.e. response.headers) to see
server metadata.
* Identify one header (e.g. Content-Type or Cache-Control) and explain what it
means

```python
print("Response Headers (Server Metadata):") # Source: Google
for header, value in response.headers.items():
    print(f"{header}: {value}")

# Cache-Control is the maximum amount of period that a server or the website can cached (store history). Some cache-controls are meant to be shared across server or browsers. 
# Some even has private cache for a user's browser ONLY.
```

Find all anchor tags
* Use soup.find_all('a') to get all < a > tags on the page.
* Print how many anchor tags were found

```python
anchor_tags = soup.find_all('a')
print(len(anchor_tags))
```

Filter for IP addresses
* Use a regular expression to match IPv4 addresses

```python
import re

ipv4_reg = r"\b(?:25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)(?:\.(?:25[0-5]|2[0-4]\d|1\d\d|[1-9]?\d)){3}\b"
links = soup.find_all("a", class_="bnone")

ips = [ip for link in links for ip in re.findall(ipv4_reg, link.text)]

print(f"IPv4 addresses: {len(ips)}")
for ip in ips:
    print(ip)

```
